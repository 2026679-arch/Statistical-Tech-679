## Data Preparation 
The analysis uses economic activity and employment size as categorical variables. The numerical measures selected are enterprise births, persons engaged at birth and enterprise surviving one year.

```python
# Selected the three numerical measures for the analysis.

selected_measures = ["Enterprise births in reference year", "Persons engaged by enterprise births in reference year", "Enterprise births surviving one year to reference year"]

# Keep the selected meausres without changing the original dataset
working_data =df.loc[df["Statistic Label"].isin(selected_measures)].copy()

#Check and verify all three measures
print("Selected measures:")
for measure in working_data["Statistic Label"].unique():
    print ("-", measure)

print("\nNumber of selected measures:", working_data["Statistic Label"].nunique())
print("Number of records:", len(working_data))
```

```python
# Display the names currently entered in the selection
print("CURRENT SELECTION:")
for label in selected_measures:
    print(repr(label))

# Display the matching survival measure from the original file
print("\nONE-YEAR SURVIVAL LABEL IN THE DATASET:")
for label in df["Statistic Label"].unique():
    if "surviving one year" in label.lower():
        print(repr(label))
```

### Selecting Consistent Categories

Broad economic sectors and individual employment-size
categories are retained to avoid overlapping totals
and subcategories.

```python
# Identify broad economic sectors
sector_endings = ("(B)", "(C)", "(D)", "(E)", "(F)", "(G)","(H)", "(I)", "(J)", "(K)", "(L)", "(M)","(N)", "(P)", "(Q)", "(R)")

broad_sectors = working_data["Activity"].str.endswith(sector_endings,na=False)
```

```python
# Exclude the combined employment-size total
individual_sizes = (working_data["Employment Size"]!= "All persons engaged size classes")
```

```python
# Create the selected dataset
prepared_data = working_data.loc[broad_sectors & individual_sizes].copy()
```

```python
# Check the selection
print("Number of sectors:", prepared_data["Activity"].nunique())
print("Number of size categories:",prepared_data["Employment Size"].nunique())

print("Number of records:", len(prepared_data))
```
